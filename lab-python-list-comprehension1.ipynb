```python
products =[ "t-shirt", "mug", "hat", "book", "keychain"]
```

```python
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    return  inventory
```

```python
inventory=initialize_inventory(products)
```

```python
def check_product_in_products():
    product_name = ""
    while(product_name not in products):
        product_name = input(f"please choose a product from {products}")
    return product_name
def get_customer_orders():
    number_product=int(input("enter the number of products"))
    name_product=[check_product_in_products()  for i in range(number_product)]
    return name_product
customer_orders=get_customer_orders()   
    
```

```python
def calculate_price():
    product_price={name_product:float(input("please enter the price of name_product"))for name_product in customer_orders}
    total=sum(product_price.values())
    return total
price=calculate_price()
```

```python
from collections import Counter
product_count=dict(Counter(customer_orders))
product_count
```

```python
update_inventory={product: inventory[product] - product_count[product] if product in product_count else inventory[product] for product in inventory.keys()}
update_inventory
```

```python
update_inventory={product:update_inventory[product] for product in update_inventory  if update_inventory[product]>0}
update_inventory
```

```python
print("Updated Inventory:")
for key, value in update_inventory.items():
    print(f"{key}:{value}")
print(f"Total Price:{price}")
```
